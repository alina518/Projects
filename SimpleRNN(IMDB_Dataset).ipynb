```python
from tensorflow.keras.preprocessing.text import one_hot
```

```python
sent = [
    'the glass of milk',
    'the glass of juice',
    'the cup of tea',
    'I am a good boy',
    'I am a good developer',
    'understand the meaning of words',
    'your videos are good',
]
```

```python
#define the vocabulary size
voc_size=10000
```

```python
#one hot representation
one_hot_repr=[one_hot(words,voc_size) for words in sent]
one_hot_repr
```

```python
#word embedding representation
from tensorflow.keras.layers import Embedding
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.models import Sequential
```

```python
import numpy as np
sent_length=8
embedding_docs=pad_sequences(one_hot_repr,padding='pre',maxlen=sent_length)
embedding_docs
```

```python
#feature representation
dim=10
model=Sequential()
model.add(Embedding(voc_size,dim,input_length=sent_length))
model.compile('adam','mse')
```

```python
model.summary()
```

```python
model.predict(embedding_docs)
```

# **End to End Deep Learning Project Using RNN**

```python
import numpy as np
import tensorflow as tf
from tensorflow.keras.datasets import imdb
from tensorflow.keras.preprocessing import sequence
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding,SimpleRNN,Dense
```

```python
## load the dataset
max_features=10000 #vocabulary size
(x_train,y_train),(x_test,y_test)=imdb.load_data(num_words=max_features)
print(x_train.shape)
print(y_train.shape)
print(x_test.shape)
print(y_test.shape)
```

```python
y_train[0]
```

```python
#inspect sample review and its label
sample_review=x_train[0]
print(sample_review)
sample_label=y_train[0]
print('Sample Label: ',sample_label)
```

```python
#mapping of words index back to words(for understanding)
word_index=imdb.get_word_index()
```

```python
reverse_word_index={value:key for key,value in word_index.items()}
reverse_word_index
```

```python
decoded_review=' '.join([reverse_word_index.get(i-3, '?') for i in sample_review])
decoded_review
```

```python
max_len=500
x_train=sequence.pad_sequences(x_train,maxlen=max_len)
x_test=sequence.pad_sequences(x_test,maxlen=max_len)
x_train
```

```python
#training simple rnn
model=Sequential()
model.add(Embedding(max_features,128,input_length=max_len))
model.add(SimpleRNN(128,activation='relu'))
model.add(Dense(1,activation='sigmoid'))
```

```python
model.summary()
```

```python
#create an instance of early stopping callback
from tensorflow.keras.callbacks import EarlyStopping
early_stopping=EarlyStopping(monitor='val_loss',patience=5,restore_best_weights=True)
early_stopping
```

```python
model.compile(optimizer='adam',loss='binary_crossentropy',metrics=['accuracy'])

history=model.fit(
    x_train,y_train,epochs=10,batch_size=32,
    validation_split=0.2,
    callbacks=early_stopping
)
```

```python
#save model file
model.save('simple_rnn_imdb.h5')
```

**Using the saved model**

```python
import numpy as np
import tensorflow as tf
from tensorflow.keras.datasets import imdb
from tensorflow.keras.preprocessing import sequence
from tensorflow.keras.models import load_model
```

```python
word_index=imdb.get_word_index()
reverse_word_index={value:key for key,value in word_index.items()}
```

```python
model=load_model('simple_rnn_imdb.h5')
model.summary()
```

```python
#helper functions to decode the reviews
def decode_review(encoded_review):
  return ' '.join(reverse_word_index.get(i-3,'?') for i in encoded_review)

#function to preprocess user input-padding
def preprocess_text(text):
  words=text.lower().split()
  encoded_review=[word_index.get(word,2)+3 for word in words]
  padded_review=sequence.pad_sequences([encoded_review],maxlen=500)
  return padded_review
```

```python
#prediction function
def predict_sentiment(review):
  preprocessed=preprocess_text(review)
  prediction=model.predict(preprocessed)
  sentiment='Positive' if prediction[0][0]>0.5 else 'Negative'
  return sentiment,prediction[0][0]
```

```python
#user input and prediction
example_review='This movie was really good! The plot was amazing and the acting was crazy good!'
sentiment,score=predict_sentiment(example_review)
print(f'Review: {example_review}')
print(f'Sentiment: {sentiment}')
print(f'Prediction Score: {score}')
```
