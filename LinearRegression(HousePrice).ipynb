##Simple Linear Regression


```python
import pandas as pd
import numpy as np

df = pd.read_csv("house_prices_dataset.csv")

target = "price"

# Fixed order
features = [
    "square_feet",
    "num_rooms",
    "age",
    "distance_to_city(km)"
]

Y = df[target].values.reshape(-1, 1)

def compute_r2(feature_list):

    X = df[feature_list].values

    # Add intercept column
    X = np.column_stack((np.ones(len(X)), X))

    beta = np.linalg.inv(X.T @ X) @ X.T @ Y

    Y_pred = X @ beta

    SSE = np.sum((Y - Y_pred) ** 2)
    SST = np.sum((Y - np.mean(Y)) ** 2)

    return 1 - (SSE / SST)

print("=" * 60)
print("INDIVIDUAL FEATURE R²")
print("=" * 60)

for feature in features:
    r2 = compute_r2([feature])
    print(f"{feature:<25} R² = {r2:.6f}")

print("\n")
print("=" * 60)
print("ADDING FEATURES ONE BY ONE")
print("=" * 60)

selected = []

for feature in features:

    selected.append(feature)

    r2 = compute_r2(selected)

    print(f"\nFeatures Used: {selected}")
    print(f"R² = {r2:.6f}")
```

##Multiple Linear Regression

```python
import pandas as pd
import numpy as np

df = pd.read_csv("house_prices_dataset.csv")

target = "price"

features = [
    "square_feet",
    "num_rooms",
    "age",
    "distance_to_city(km)"
]

# X matrix
X = df[features].values

# add intercept column
X = np.column_stack((np.ones(len(X)), X))

# Y vector
Y = df[target].values.reshape(-1,1)

# Normal Equation
beta = np.linalg.inv(X.T @ X) @ X.T @ Y

print("="*70)
print("MULTIPLE LINEAR REGRESSION")
print("="*70)

print("\nCoefficients:\n")

print(f"b0 = {beta[0,0]:.4f}")

for i, feature in enumerate(features):
    print(f"b{i+1} ({feature}) = {beta[i+1,0]:.4f}")

# Predictions
Y_pred = X @ beta

# R²
sse = np.sum((Y - Y_pred)**2)
sst = np.sum((Y - np.mean(Y))**2)

r2 = 1 - (sse/sst)

print(f"\nR² = {r2:.6f}")

print("\nRegression Equation:\n")

equation = f"Price = {beta[0,0]:.4f}"

for i, feature in enumerate(features):
    equation += f" + ({beta[i+1,0]:.4f})*{feature}"

print(equation)
```

```python
import pandas as pd
import numpy as np

# Load dataset
df = pd.read_csv("house_prices_dataset.csv")

# Features
X = df[
    [
        "square_feet",
        "num_rooms",
        "age",
        "distance_to_city(km)"
    ]
].values

# Target
y = df["price"].values.reshape(-1,1)

# Feature Scaling (important for GD)
X_mean = X.mean(axis=0)
X_std = X.std(axis=0)

X = (X - X_mean) / X_std

# Add intercept column
X = np.column_stack((np.ones(len(X)), X))

# Initialize parameters
m, n = X.shape

theta = np.zeros((n,1))

# Hyperparameters
learning_rate = 0.01
epochs = 5000

# Cost Function
def compute_cost(X, y, theta):

    m = len(y)

    predictions = X @ theta

    cost = (1/(2*m)) * np.sum((predictions-y)**2)

    return cost

# Gradient Descent
cost_history = []

for epoch in range(epochs):

    predictions = X @ theta

    gradients = (1/m) * X.T @ (predictions-y)

    theta = theta - learning_rate * gradients

    cost = compute_cost(X,y,theta)

    cost_history.append(cost)

    if epoch % 500 == 0:
        print(f"Epoch {epoch} Cost = {cost:.4f}")

print("\nCoefficients")

print(f"b0 = {theta[0,0]}")

for i,col in enumerate([
        "square_feet",
        "num_rooms",
        "age",
        "distance_to_city(km)"
]):
    print(f"b{i+1} ({col}) = {theta[i+1,0]}")

# Predictions
y_pred = X @ theta

# R²
sse = np.sum((y-y_pred)**2)

sst = np.sum((y-np.mean(y))**2)

r2 = 1 - (sse/sst)

print(f"\nR² = {r2:.6f}")
```
